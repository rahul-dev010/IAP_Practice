## Phase 1: First DLT Pipeline (Bronze Layer)

- **Source:** `samples.tpch` tables.
- **Pipeline:** `tpch_dlt_pipeline_v1`
- **Notebook:** `dlt_bronze_tpch` (Python)
- **Target:** `dlt_retail.bronze`
- **Mode:** Triggered

### Key Concepts
- **Declarative:** Define what the table should contain; DLT manages execution and dependencies.
- **`@dlt.table`:** Defines a pipeline-managed table.
- **Lineage:** Tracks dependencies between tables automatically.
- **Compute:** The pipeline manages its own compute.

### Bronze Tables
- `bronze_customer` → `samples.tpch.customer`
- `bronze_orders` → `samples.tpch.orders`
- `bronze_lineitem` → `samples.tpch.lineitem`
- `bronze_part` → `samples.tpch.part`
