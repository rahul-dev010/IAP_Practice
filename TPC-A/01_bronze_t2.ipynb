```python
catalog = dbutils.widgets.get("catalog")
```

```python
bronze_tbl = f"{catalog}.bronze.airline"

# Read the raw airline CSV
raw_df = (
    spark.read.format("csv")
    .option("header", True)
    .load("/Volumes/tpca/bronze/raw/airline/")
)

# Append metadata columns (_source_file, _ingested_at, pipeline_run_id)
airline_df = add_metadata(raw_df)

airline_df.createOrReplaceTempView("airline_staged")

spark.sql(f"""
CREATE OR REPLACE TABLE {bronze_tbl} AS
SELECT * FROM airline_staged
""")

# print(f"Created table: {bronze_tbl}")
```
